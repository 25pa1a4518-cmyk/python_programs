# **LIST OPERATIONS**

```python
L1=[1,2,3,4,5,6,7,8,"hello"]
L2=[3,7,8,"sanjana"]
#CONCATENATION
print(L1+L2)
```

```python
#REPETITION
print(L1*3)
```

```python
#INDEXING
print(L2[3])
```

```python
#SLICING
print(L1[0:3])
print(L1[0:5:2])
print(L1[::-1]) #reverse of a list
```

```python
#MEMBERSHIP OPERATOR
if(7 in L1):
  print("true")
else:
  print("false")
```

```python
#ITERATION
for i in L2:
  print(i)
for i in range(len(L2)):
  print(L2[i])
```

# **LIST FUNCTIONS**

```python
L3=[25,67,89,35,43,21,98,35,64]
print(max(L3))
print(min(L3))
print(sum(L3))
print(sorted(L3))
print(list(reversed(L3)))
print(L3)
```

# **LIST METHODS**

```python
L4=[89,56,78,23,42,56,90]
L5=[8,18,28,38,48,58]
#APPEND
L4.append(12)
print(L4)
#EXTEND
L5.extend(L4)
print(L5)
#INSERT
L4.insert(6,"hello")
print(L4)
#POP
L4.pop()
L4.pop(6)
print(L4)
#REMOVE
L5.remove(8)
print(L5)
#SORT
L5.sort()
print(L5)
#REVERSE
L5.reverse()
print(L5)

```

```python
#program that prints sum of list elements without using sum()
L6=[25,67,89,45,67,34]
sum=0
for i in L6:
  sum=sum+i
print(sum)

```

```python
#print the list elements at odd index
L7=[25,67,89,45,67,34]
for i in range(1,len(L7),2):
  print(L7[i])
```

```python
#print the even elements
L8=[25,67,89,45,67,34]
for i in L8:
  if(i%2==0):
    print(i)
```
