```python
with open("new.txt","r")as file:
  print(file.readline())
  print(file.tell())
  file.seek(1)
  print(file.tell())
  print(file.readlines())
```

```python
#copy data in file to another
with open("new.txt","r") as file:
  data=file.read()
with open("new2.txt","w") as f:
  f.write(data)
with open("new2.txt","r") as f:
  print(f.readlines())

```

```python
#print each line of a file in reverse order
with open("new.txt","r") as file:
  data=file.readlines()
  for line in data[::-1]:
    print(line)
```

```python
#Write a program to sort words in a file and put them in another file. The output file should have only lower-case words only
with open("f1.txt","w") as file:
  file.writelines(["\napple","\nname","\nend","\nsanjana","\nhello"])
with open("f1.txt","r") as file:
  data=file.read().split()
  d=[i.lower() for i in data]
  d.sort()
with open("f2.txt","w") as f:
  for i in d:
    f.write(i+"\n")
with open("f2.txt","r") as f:
  print(f.readlines())
```

# ***BINARY FILE HANDLING***

```python
#rb,wb,ab
with open("picture1.jpeg","rb") as f:
  data=f.read()
  print(data)
with open("picture2.jpeg","wb") as f2:
  f2.write(data)
print("pic created")
```

```python
import pickle
with open("s1.data","wb") as f:
  data=[1,2,3,"hello"]
  pickle.dump(data,f)
with open("s1.data","rb") as f:
  data=pickle.load(f)
  print(data)
```

```python
f = open("sample.txt", "w")
f.write("Hello Python\nWelcome to Programming")
f.close()

f = open("sample.txt", "r")
content = f.read()
f.close()

characters = len(content)
words = len(content.split())
lines = len(content.splitlines())

print("Characters:", characters)
print("Words:", words)
print("Lines:", lines)
```
