```python
#ARITHMETIC OPERATORS
a=int(input("enter an integer:"))
b=int(input("enter another integer:"))
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a**b)
print(a//b)

```

```python
#COMPARISON OPERATORS
a=int(input("enter an integer:"))
b=int(input("enter another integer:"))
print(a==b)
print(a!=b)
print(a>b)
print(a<b)
print(a>=b)
print(a<=b)

```

```python
#LOGICAL OPERATORS
a=int(input("enter an integer:"))
b=int(input("enter another integer:"))
print(a==b and a!=b)
print(a==b or a!=b)
print(not a==b)
```
