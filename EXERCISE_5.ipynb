```python
#REVERSE A STRING USING FUNCTIONS
def reverse(string):
    s = string[::-1]
    return s
reverse("python")
```

```python
#REVERSE A STRING USING FUNCTIONS AND LOOPS
def rev(string):
    rev= ""
    for i in string:
        rev = i + rev
    return rev
rev("PYTHONPROGRAMMING")
```

```python
#example code on keyword arguments
def subtraction(a,b,c):
  print("the values are",a,b,c)
  return a-b-c
res=subtraction(a=6,b=7,c=9)
print(res)
print(subtraction(b=89,a=76,c=34))
print(subtraction(65,54,32))
```

```python
#function that return multiple values
def multiple(*num):
  s=sum(num)
  c=len(num)
  avg=s/c
  return s,c,avg
a,b,c=multiple(1,2,3,4,5,6,7,8,9,67,54,89,7645)
print(f"the sum of numbers is {a},the count is {b},the average is {b:.2f}")
```
